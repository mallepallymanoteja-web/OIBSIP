```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("/crime_incidents_messy.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("First 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# Count duplicates before cleaning
duplicate_count = df.duplicated().sum()

print("Duplicate rows found:", duplicate_count)

# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("New dataset shape:", df.shape)
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values[missing_values > 0])
```

```python
missing_summary = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': (df.isnull().sum() / len(df) * 100).round(2)
})

missing_summary = missing_summary[missing_summary['Missing Count'] > 0]
missing_summary = missing_summary.sort_values('Missing Percentage', ascending=False)

display(missing_summary)
```

```python
categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).head(15))
```

```python
# Clean text columns by removing extra spaces
# and standardizing capitalization

text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip().title() if isinstance(x, str) else x
    )

print("Text columns cleaned successfully!")
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numeric columns:")
print(list(numeric_columns))

print("\nBasic statistics:")
display(df[numeric_columns].describe().T)
```

```python
print("Suspect age range:")
print("Minimum:", df['suspect_age'].min())
print("Maximum:", df['suspect_age'].max())

print("\nVictim age range:")
print("Minimum:", df['victim_age'].min())
print("Maximum:", df['victim_age'].max())
```

```python
# Replace impossible ages with missing values
# Valid age range considered: 0 to 100

df.loc[(df['suspect_age'] < 0) | (df['suspect_age'] > 100), 'suspect_age'] = np.nan
df.loc[(df['victim_age'] < 0) | (df['victim_age'] > 100), 'victim_age'] = np.nan

print("Invalid age values have been replaced with missing values.")

print("\nSuspect age range after cleaning:")
print("Minimum:", df['suspect_age'].min())
print("Maximum:", df['suspect_age'].max())

print("\nVictim age range after cleaning:")
print("Minimum:", df['victim_age'].min())
print("Maximum:", df['victim_age'].max())
```

```python
print("Latitude range:")
print("Minimum:", df['latitude'].min())
print("Maximum:", df['latitude'].max())

print("\nLongitude range:")
print("Minimum:", df['longitude'].min())
print("Maximum:", df['longitude'].max())
```

```python
# Replace invalid latitude values with missing values
df.loc[(df['latitude'] < -90) | (df['latitude'] > 90), 'latitude'] = np.nan

print("Invalid latitude values have been replaced with missing values.")

print("\nLatitude range after cleaning:")
print("Minimum:", df['latitude'].min())
print("Maximum:", df['latitude'].max())
```

```python
print("Numeric columns and their ranges:\n")

for col in df.select_dtypes(include=np.number).columns:
    print(f"{col}: {df[col].min()} to {df[col].max()}")
```

```python
# Replace negative arrest counts with missing values
df.loc[df['num_arrests'] < 0, 'num_arrests'] = np.nan

print("Invalid negative arrest counts have been replaced with missing values.")

print("\nArrest count range after cleaning:")
print("Minimum:", df['num_arrests'].min())
print("Maximum:", df['num_arrests'].max())
```

```python
categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).head(20))
```

```python
# Standardize suspect gender
df['suspect_gender'] = df['suspect_gender'].replace({
    'M': 'Male',
    'F': 'Female'
})

# Standardize victim gender
df['victim_gender'] = df['victim_gender'].replace({
    'M': 'Male',
    'F': 'Female'
})

print("Suspect gender values after cleaning:")
print(df['suspect_gender'].value_counts(dropna=False))

print("\nVictim gender values after cleaning:")
print(df['victim_gender'].value_counts(dropna=False))
```

```python
# Standardize abbreviated district names
df['district'] = df['district'].replace({
    'Nor': 'North',
    'Sou': 'South',
    'Cen': 'Central',
    'Mid': 'Midtown',
    'Eas': 'East',
    'Wes': 'West'
})

print("District values after cleaning:")
print(df['district'].value_counts(dropna=False))
```

```python
# Standardize inconsistent crime type names
df['crime_type'] = df['crime_type'].replace({
    'Dv': 'Domestic Violence',
    'Drug Offence': 'Drug Offense'
})

print("Crime types after cleaning:")
print(df['crime_type'].value_counts(dropna=False))
```

```python
# Remove extra spaces from crime type names
df['crime_type'] = df['crime_type'].apply(
    lambda x: ' '.join(x.split()) if isinstance(x, str) else x
)

# Standardize obvious variants
df['crime_type'] = df['crime_type'].replace({
    'Drug Offence': 'Drug Offense',
    'Dv': 'Domestic Violence'
})

print("Crime types after final cleaning:")
print(df['crime_type'].value_counts(dropna=False))
```

```python
# Standardize case status values
df['case_status'] = df['case_status'].replace({
    'Investgation': 'Under Investigation',
    'Pendng': 'Pending'
})

print("Case status after cleaning:")
print(df['case_status'].value_counts(dropna=False))
```

```python
# Standardize resolution values
df['resolution'] = df['resolution'].replace({
    'Arres Made': 'Arrest Made',
    'Warning Issued': 'Warning',
    'Case Dismissed': 'Dismissed'
})

print("Resolution after cleaning:")
print(df['resolution'].value_counts(dropna=False))
```

```python
# Standardize severity values
df['severity'] = df['severity'].replace({
    '1': 'Low',
    '2': 'Medium',
    '3': 'High',
    '4': 'Critical',
    'Med': 'Medium',
    'Crit': 'Critical',
    1: 'Low',
    2: 'Medium',
    3: 'High',
    4: 'Critical'
})

print("Severity after cleaning:")
print(df['severity'].value_counts(dropna=False))
```

```python
# Standardize reported_online values
df['reported_online'] = df['reported_online'].replace({
    'Yes': 'Yes',
    'True': 'Yes',
    '1': 'Yes',
    1: 'Yes',
    'No': 'No',
    'False': 'No',
    '0': 'No',
    0: 'No'
})

print("Reported online after cleaning:")
print(df['reported_online'].value_counts(dropna=False))
```

```python
# Convert incident_datetime to proper datetime format
df['incident_datetime'] = pd.to_datetime(
    df['incident_datetime'],
    errors='coerce'
)

print("Data type:", df['incident_datetime'].dtype)
print("Missing/invalid dates:", df['incident_datetime'].isna().sum())

print("\nDate range:")
print("Earliest:", df['incident_datetime'].min())
print("Latest:", df['incident_datetime'].max())
```

```python
def clean_phone(value):
    if pd.isna(value):
        return np.nan

    digits = ''.join(ch for ch in str(value) if ch.isdigit())

    if len(digits) == 10:
        return digits
    else:
        return np.nan

df['victim_phone'] = df['victim_phone'].apply(clean_phone)

print("Valid phone numbers:", df['victim_phone'].notna().sum())
print("Missing/invalid phone numbers:", df['victim_phone'].isna().sum())

print("\nSample cleaned phone numbers:")
print(df['victim_phone'].dropna().head(10))
```

```python
# Standardize weapon names
df['weapon_used'] = df['weapon_used'].replace({
    'Hands': 'Hands/Feet',
    'Gun': 'Firearm'
})

print("Weapon types after cleaning:")
print(df['weapon_used'].value_counts(dropna=False))
```

```python
print("========== FINAL DATA VALIDATION ==========\n")

print("Dataset shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())

print("\nRemaining missing values:")
missing = df.isnull().sum()
print(missing[missing > 0])

print("\nAge validation:")
print("Suspect age:", df['suspect_age'].min(), "to", df['suspect_age'].max())
print("Victim age:", df['victim_age'].min(), "to", df['victim_age'].max())

print("\nLatitude validation:")
print("Latitude:", df['latitude'].min(), "to", df['latitude'].max())

print("\nArrest count validation:")
print("Arrests:", df['num_arrests'].min(), "to", df['num_arrests'].max())

print("\nDate validation:")
print("Invalid/missing dates:", df['incident_datetime'].isna().sum())

print("\nCleaning validation completed successfully!")
```

```python
# Save the cleaned dataset
df.to_csv("/cleaned_crime_incidents.csv", index=False)

print("Cleaned dataset saved successfully!")
print("File: cleaned_crime_incidents.csv")
print("Final shape:", df.shape)
```
